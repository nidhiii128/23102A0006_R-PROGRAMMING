```r
x = c(2,4,6,8,10,12)

excount = function(x){
    count = 0
    for (xval in x) {
        if (xval/2 > 3)
            count = count + 1
    }
    print(count)
}

excount(x)
```

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
    if (x == "tea") {
        break
    }
    print(x)
}
```

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
    if (x == "lemonade") {
        next
    }
    print(x)
}
```

```r
i = 1

while (i < 10) {
    print(i^2)
    i = i + 2
}
```

```r
sumfunction = function() {

    sum = 0

    number = as.integer(
        readline(prompt = "Please select any number less than 25: ")
    )

    while (number <= 25) {
        sum = sum + number
        number = number + 1
    }

    print(paste("The sum of numbers received from the While Loop:", sum))
}

sumfunction()
```

```r
i = 1

repeat {

    print(i^2)

    i = i + 2

    if (i > 10)
        break
}
```

```r
i = 1

repeat {

    i = i + 1

    if (i < 10)
        next

    print(i^2)

    if (i >= 13)
        break
}
```
