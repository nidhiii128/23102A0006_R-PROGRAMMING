```r
missing_summary <- function(data){

vars <- c("PM2.5","PM10","SO2","NO2","TEMP","WSPM","wd")

result <- data.frame(
Variable=character(),
Total_Records=integer(),
Missing_Values=integer(),
Missing_Percentage=numeric()
)

for(i in vars){

total <- nrow(data)
missing <- sum(is.na(data[[i]]))
percent <- (missing/total)*100

result <- rbind(result,
data.frame(
Variable=i,
Total_Records=total,
Missing_Values=missing,
Missing_Percentage=round(percent,2)
))

if(percent>20){
warning(paste(i,"has more than 20% missing values"))
}

}

return(result)

}

missing_summary(air)
```

```r
num_cols <- c("PM2.5","PM10","SO2","NO2","CO","O3","TEMP","PRES","DEWP","RAIN","WSPM")

for(col in num_cols){

mean_value <- mean(air[[col]],na.rm=TRUE)

air[[col]][is.na(air[[col]])] <- mean_value

}
```

```r
mode_fun <- function(x){

ux <- unique(x)

ux[which.max(tabulate(match(x,ux)))]

}

air$wd[is.na(air$wd)] <- mode_fun(air$wd)
```

```r
clean_variable <- function(data,column){

tryCatch({

if(!(column %in% names(data))){
stop("Column not found")
}

if(is.numeric(data[[column]])){

data[[column]][is.na(data[[column]])] <-
mean(data[[column]],na.rm=TRUE)

}else{

mode_value <- mode_fun(data[[column]])
data[[column]][is.na(data[[column]])] <- mode_value

}

return(data)

},
error=function(e){

print(e$message)

})

}
```

```r
air <- clean_variable(air,"PM2.5")
```

```r
comparison <- data.frame(

Method=c("Mean","Median","Mode"),

PM2.5=c(
mean(air$PM2.5),
median(air$PM2.5),
mode_fun(air$PM2.5)
),

TEMP=c(
mean(air$TEMP),
median(air$TEMP),
mode_fun(air$TEMP)
)

)

comparison
```
